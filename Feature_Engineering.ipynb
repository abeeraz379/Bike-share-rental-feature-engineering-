# Apply Feature Engineering

## Load Data

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/drive/MyDrive/AXSOSACADEMY/AXSOSACADEMY/05-IntermediateML/Week17/Data/bikeshare_train - bikeshare_train.csv')
```

```python
df.head()
```

```python
df.info()
```

##  drop 'casual' and 'registered' columns and target 'count'

```python
df=df.drop(['casual','registered','count'],axis=1)
df.info()
```

##  Transform the 'datetime' column into a datetime type and use it to create 3 new columns

```python
# Transform the 'datetime' column into a datetime type and use it to create 3 new columns in 1. Name of the Month 2. Name of the Day of the Week 3. Hour of the Day
df['datetime']=pd.to_datetime(df['datetime'])
df['month']=df['datetime'].dt.month
df['day']=df['datetime'].dt.day
df['hour']=df['datetime'].dt.hour
```

```python
df.info()
```

```python
# change month , day , hour columns to object so we can apply onehotencoder
df['month']=df['month'].astype(str)
df['day']=df['day'].astype(str)
df['hour']=df['hour'].astype(str)
```

## Drop the 'datetime' and 'season' columns. These are now redundant.

```python
# Drop the 'datetime' and 'season' columns.
df=df.drop(['datetime','season'],axis=1)
df.info()
```

## The temperatures in the 'temp' and 'atemp' columns are in Celsius. Use '.apply() and a Lambda function to convert them to Fahrenheit.

```python
# The temperatures in the 'temp' and 'atemp' columns are in Celsius. Use '.apply() and a Lambda function to convert them to Fahrenheit.
df['temp']=df['temp'].apply(lambda x: (x * 9/5) + 32)
df['atemp']=df['atemp'].apply(lambda x: (x * 9/5) + 32)
```

## Create a new column, 'temp_variance,' which shows how much warmer or colder the current temperature ('temp') is than the average temperate for that day of the year ('atemp'). If the current temperature is warmer than average ('atemp'), the value in 'temp_variance' should be positive.

```python
# Create a new column, 'temp_variance,' which shows how much warmer or colder the current temperature ('temp') is than the average temperate for that day of the year ('atemp'). If the current temperature is warmer than average ('atemp'), the value in 'temp_variance' should be positive.
df['temp_variance']=df['temp']-df['atemp']
```

```python
# drop atemp column
df=df.drop(['atemp'],axis=1)
```

## The features after feature engineering

```python
df.info()
```

```python
new_df=df
```

```python
new_df.info()
```

# Model with feature Engineering

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score , mean_absolute_error
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
# make _pipeline
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv('/content/drive/MyDrive/AXSOSACADEMY/AXSOSACADEMY/05-IntermediateML/Week17/Data/bikeshare_train - bikeshare_train.csv')
```

```python
# define x and y
x=new_df
y=df["count"]
```

```python

# split data
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=1)
```

```python

# impute missing value with median
# build scaler
# build one hot encoder

num_cols = x_train.select_dtypes("number").columns
cat_cols = x_train.select_dtypes("object").columns

num_transformer = make_pipeline(SimpleImputer(strategy='median'), StandardScaler())
cat_transformer = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore'))
preprocessor = ColumnTransformer([('num', num_transformer, num_cols), ('cat', cat_transformer, cat_cols)])

```

```python
lr=LinearRegression()
pipe=make_pipeline(preprocessor,lr)
pipe.fit(x_train,y_train)
```

```python
y_pred=pipe.predict(x_test)
print("MSE:", mean_squared_error(y_test,y_pred))
print("MAE:", mean_absolute_error(y_test,y_pred))
print("R2:", r2_score(y_test,y_pred))
```
